```python
import pandas as pd
import numpy as np

from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression

import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

!pip install optbinning -q
from optbinning import OptimalBinning
```

```python
datos = pd.read_parquet("base_analytics_lab_061628.parquet")
datos.head()
```

```python
TARGET = "TARGET"
COL_MES = "CODMES"

print(datos.shape)
print("Duplicados:", datos.duplicated().sum())

resumen = datos.groupby(COL_MES).agg(
    n=(TARGET, "size"),
    malos=(TARGET, "sum"),
    tasa_12m=(TARGET, "mean"),
    tasa_6m=("PROXY_TARGET_6M", "mean"),
)
resumen["gini_modelo_anterior"] = datos.groupby(COL_MES).apply(
    lambda d: 2 * roc_auc_score(d[TARGET], d["PREDICCION_MODELO_ANTERIOR"].fillna(0)) - 1
)
print(resumen)

nulos = datos.isna().mean().sort_values(ascending=False)
print(nulos[nulos > 0])
```

```python
meses = sorted(datos[COL_MES].unique())
TRAIN_MESES = meses[:-6]
VALID_MESES = meses[-6:-3]
OOT_MESES   = meses[-3:]

datos["PARTICION"] = np.select(
    [datos[COL_MES].isin(TRAIN_MESES), datos[COL_MES].isin(VALID_MESES)],
    ["TRAIN", "VALID"],
    default="OOT",
)

print(datos.groupby("PARTICION")[TARGET].agg(["size", "sum", "mean"]))
```

```python
FEATURES_EXCLUIR = [COL_MES, TARGET, "PROXY_TARGET_6M",
                    "PREDICCION_MODELO_ANTERIOR", "PARTICION"]
FEATURES = [c for c in datos.columns if c not in FEATURES_EXCLUIR]

train = datos[datos["PARTICION"] == "TRAIN"]
valid = datos[datos["PARTICION"] == "VALID"]

def iv_con_bins(bins, y):
    t = pd.crosstab(bins, y).reindex(columns=[0, 1], fill_value=0) + 0.5
    dg, db = t[0] / t[0].sum(), t[1] / t[1].sum()
    return ((dg - db) * np.log(dg / db)).sum()

def psi(base, nueva):
    p, q = base.value_counts(normalize=True).align(
        nueva.value_counts(normalize=True), fill_value=0)
    p, q = p + 1e-6, q + 1e-6
    return ((q - p) * np.log(q / p)).sum()

binners, filas = {}, []
for var in FEATURES:
    dtype = "categorical" if train[var].dtype == "object" else "numerical"
    ob = OptimalBinning(name=var, dtype=dtype, solver="cp",
                        monotonic_trend="auto_asc_desc",
                        min_bin_size=0.05, max_n_bins=6)
    ob.fit(train[var].values, train[TARGET].values)
    ob.binning_table.build()
    binners[var] = ob

    b_tr = pd.Series(ob.transform(train[var].values, metric="bins"), index=train.index)
    b_va = pd.Series(ob.transform(valid[var].values, metric="bins"), index=valid.index)

    iv_mes = [iv_con_bins(b_tr[train[COL_MES] == m], train.loc[train[COL_MES] == m, TARGET])
              for m in TRAIN_MESES]

    filas.append({
        "variable": var,
        "n_bins": len(ob.splits) + 1,
        "iv_train": ob.binning_table.iv,
        "iv_min_mes": np.min(iv_mes),
        "cv_iv": np.std(iv_mes) / np.mean(iv_mes),
        "csi_valid": psi(b_tr, b_va),
    })

diag = pd.DataFrame(filas).sort_values("iv_train", ascending=False)
diag["flag"] = np.select(
    [diag["iv_train"] < 0.02, diag["csi_valid"] > 0.25, diag["cv_iv"] > 0.3, diag["csi_valid"] > 0.1],
    ["IV bajo", "CSI alto", "IV inestable", "CSI moderado"],
    default="OK",
)
print(diag.round(4).to_string(index=False))
```

```python
def direccion(ob):
    t = ob.binning_table.build()
    t = t[(t.index != "Totals") & (~t["Bin"].astype(str).isin(["Special", "Missing"]))]
    er = t["Event rate"].values
    if len(er) < 2:
        return "plano"
    return "ascendente" if er[-1] > er[0] else "descendente"

diag["direccion_datos"] = diag["variable"].map(lambda v: direccion(binners[v]))
print(diag[["variable", "iv_train", "direccion_datos"]].to_string(index=False))
```

```python
candidatas = diag.loc[diag["iv_train"] >= 0.02, "variable"].tolist()

for max_bins in [4, 6, 8, 10]:
    iv_tr, iv_va = [], []
    for var in candidatas:
        dtype = "categorical" if train[var].dtype == "object" else "numerical"
        ob = OptimalBinning(name=var, dtype=dtype, solver="cp",
                            monotonic_trend="auto_asc_desc",
                            min_bin_size=0.05, max_n_bins=max_bins)
        ob.fit(train[var].values, train[TARGET].values)
        b_va = pd.Series(ob.transform(valid[var].values, metric="bins"), index=valid.index)
        ob.binning_table.build()
        iv_tr.append(ob.binning_table.iv)
        iv_va.append(iv_con_bins(b_va, valid[TARGET]))
    print(f"max_bins={max_bins}: IV_train medio={np.mean(iv_tr):.4f} | "
          f"IV_valid medio={np.mean(iv_va):.4f} | caída={1 - np.mean(iv_va)/np.mean(iv_tr):.1%}")
```

```python
dev = datos[datos["PARTICION"].isin(["TRAIN", "VALID"])]
DEV_MESES = list(TRAIN_MESES) + list(VALID_MESES)

def n_bins_reales(ob):
    t = ob.binning_table.build()
    return ((t.index != "Totals") & (~t["Bin"].astype(str).isin(["Special", "Missing"]))).sum()

filas = []
for var in candidatas:
    ob = binners[var]
    nb = n_bins_reales(ob)
    signo = 1 if diag.set_index("variable").loc[var, "direccion_datos"] == "ascendente" else -1
    idx = pd.Series(ob.transform(dev[var].values, metric="indices"), index=dev.index)

    ivs, rhos = [], []
    for m in DEV_MESES:
        mask = dev[COL_MES] == m
        ivs.append(iv_con_bins(idx[mask], dev.loc[mask, TARGET]))
        er = dev.loc[mask].groupby(idx[mask])[TARGET].mean()
        er = er[er.index < nb]
        rhos.append(signo * er.index.to_series().corr(er, method="spearman") if len(er) > 2 else np.nan)

    filas.append({
        "variable": var,
        "iv_primeros3": np.mean(ivs[:3]),
        "iv_ultimos3": np.mean(ivs[-3:]),
        "ratio_iv": np.mean(ivs[-3:]) / np.mean(ivs[:3]),
        "rho_min": np.nanmin(rhos),
        "meses_invertidos": int(np.sum(np.array(rhos) < 0)),
    })

estab = pd.DataFrame(filas).merge(diag[["variable", "iv_train", "direccion_datos"]], on="variable")
estab["flag_tendencia"] = np.select(
    [estab["meses_invertidos"] > 0, estab["ratio_iv"] < 0.5, estab["rho_min"] < 0.6],
    ["DIRECCION INVERTIDA", "IV CAYENDO", "ORDEN DEBIL"],
    default="OK",
)
print(estab.sort_values("iv_train", ascending=False).round(3).to_string(index=False))
```

```python
KEEP = estab.loc[estab["iv_ultimos3"] >= 0.03, "variable"].tolist()

def a_woe(df, vars_):
    return pd.DataFrame({v: binners[v].transform(df[v].values, metric="woe") for v in vars_},
                        index=df.index)

woe_tr = a_woe(train, KEEP)
corr = woe_tr.corr().abs()

orden = estab.set_index("variable").loc[KEEP, "iv_ultimos3"].sort_values(ascending=False).index
seleccion, descartes = [], []
for v in orden:
    conflicto = [s for s in seleccion if corr.loc[v, s] > 0.7]
    if conflicto:
        descartes.append({"descartada": v, "por": conflicto[0],
                          "corr": round(corr.loc[v, conflicto[0]], 2)})
    else:
        seleccion.append(v)

print(f"Seleccionadas ({len(seleccion)}):", seleccion)
print(pd.DataFrame(descartes).to_string(index=False))

X = sm.add_constant(woe_tr[seleccion])
vif = pd.Series([variance_inflation_factor(X.values, i) for i in range(1, X.shape[1])],
                index=seleccion).sort_values(ascending=False)
print(vif.round(2))
```

```python
VARS = seleccion
woe_va = a_woe(valid, VARS)
y_tr, y_va = train[TARGET], valid[TARGET]

def gini(y, p):
    return 2 * roc_auc_score(y, p) - 1

def gini_mensual(p, df):
    s = pd.Series(p, index=df.index)
    return s.groupby(df[COL_MES]).apply(lambda x: gini(df.loc[x.index, TARGET], x))

signo_uni = np.sign(woe_tr[VARS].corrwith(y_tr))
print("Signos univariados:", signo_uni.value_counts().to_dict())

def evaluar(modelo, vars_):
    p = modelo.predict_proba(woe_va[vars_])[:, 1]
    coef = pd.Series(modelo.coef_[0], index=vars_)
    activos = coef[coef != 0]
    return {
        "gini_valid": gini(y_va, p),
        "gini_min_mes": gini_mensual(p, valid).min(),
        "n_vars": len(activos),
        "signos_invertidos": int((np.sign(activos) != signo_uni[activos.index]).sum()),
    }

print("Benchmark modelo anterior en VALID:",
      round(gini(y_va, valid["PREDICCION_MODELO_ANTERIOR"].fillna(0)), 4))

# 1. Sin penalización
m_full = LogisticRegression(penalty=None, max_iter=1000).fit(woe_tr[VARS], y_tr)
print("Sin penalización:", {k: round(v, 4) for k, v in evaluar(m_full, VARS).items()})

# 2. Lasso: grilla de C
res = []
for C in np.logspace(-4, 0, 25):
    m = LogisticRegression(penalty="l1", C=C, solver="liblinear", max_iter=1000)
    m.fit(woe_tr[VARS], y_tr)
    res.append({"C": C, **evaluar(m, VARS)})
lasso = pd.DataFrame(res)
print(lasso.round(4).to_string(index=False))

# 3. Stepwise forward por Gini en VALID
elegidas, restantes, mejor = [], list(VARS), 0.0
while restantes:
    cand = {}
    for v in restantes:
        m = LogisticRegression(penalty=None, max_iter=1000).fit(woe_tr[elegidas + [v]], y_tr)
        cand[v] = gini(y_va, m.predict_proba(woe_va[elegidas + [v]])[:, 1])
    v_best = max(cand, key=cand.get)
    if cand[v_best] - mejor < 0.002:
        break
    elegidas.append(v_best)
    restantes.remove(v_best)
    mejor = cand[v_best]
    print(f"+ {v_best:<42} gini_valid={mejor:.4f}")

m_step = LogisticRegression(penalty=None, max_iter=1000).fit(woe_tr[elegidas], y_tr)
print("Stepwise:", {k: round(v, 4) for k, v in evaluar(m_step, elegidas).items()})
```

```python
p_full = m_full.predict_proba(woe_va[VARS])[:, 1]
p_step = m_step.predict_proba(woe_va[elegidas])[:, 1]
p_ant = valid["PREDICCION_MODELO_ANTERIOR"].fillna(0).values
y = y_va.values

rng = np.random.default_rng(42)
dif_step, dif_ant = [], []
for _ in range(1000):
    i = rng.integers(0, len(y), len(y))
    g_full = gini(y[i], p_full[i])
    dif_step.append(g_full - gini(y[i], p_step[i]))
    dif_ant.append(g_full - gini(y[i], p_ant[i]))

print("Full - Stepwise   IC95%:", np.percentile(dif_step, [2.5, 97.5]).round(4))
print("Full - M.anterior IC95%:", np.percentile(dif_ant, [2.5, 97.5]).round(4))
```

```python
dev = datos[datos["PARTICION"].isin(["TRAIN", "VALID"])]
oot = datos[datos["PARTICION"] == "OOT"]

def mapa_woe(idx, y):
    t = pd.crosstab(idx, y).reindex(columns=[0, 1], fill_value=0) + 0.5
    return np.log((t[0] / t[0].sum()) / (t[1] / t[1].sum()))

def indices(df, var):
    return pd.Series(binners[var].transform(df[var].values, metric="indices"), index=df.index)

mapas_train = {v: mapa_woe(indices(train, v), train[TARGET]) for v in VARS_FINAL}
mapas_dev   = {v: mapa_woe(indices(dev, v), dev[TARGET]) for v in VARS_FINAL}

def aplicar(df, mapas):
    return pd.DataFrame({v: indices(df, v).map(mapas[v]).fillna(0) for v in VARS_FINAL},
                        index=df.index)

# Chequeo de convención: WoE manual vs optbinning en TRAIN
print("Corr mínima con optbinning:",
      round(aplicar(train, mapas_train).corrwith(woe_tr[VARS_FINAL]).min(), 4))

# Chequeo de dirección: el orden del WoE por bin debe ser igual en TRAIN y en DEV
cambios = []
for v in VARS_FINAL:
    nb = n_bins_reales(binners[v])
    a = mapas_train[v][mapas_train[v].index < nb]
    b = mapas_dev[v][mapas_dev[v].index < nb]
    if len(a) > 1 and np.sign(a.iloc[-1] - a.iloc[0]) != np.sign(b.iloc[-1] - b.iloc[0]):
        cambios.append(v)
print("Variables que cambiaron de dirección:", cambios or "ninguna")

woe_dev = aplicar(dev, mapas_dev)
woe_oot = aplicar(oot, mapas_dev)
modelo_final = LogisticRegression(penalty=None, max_iter=1000).fit(woe_dev, dev[TARGET])
print(pd.Series(modelo_final.coef_[0], index=VARS_FINAL).round(4))
```

```python
p_dev = modelo_final.predict_proba(woe_dev)[:, 1]
p_oot = modelo_final.predict_proba(woe_oot)[:, 1]
p_ant_oot = oot["PREDICCION_MODELO_ANTERIOR"].fillna(0).values
y_oot = oot[TARGET].values

# 1. Discriminación
rng = np.random.default_rng(42)
g_nuevo, g_dif = [], []
for _ in range(1000):
    i = rng.integers(0, len(y_oot), len(y_oot))
    gn = gini(y_oot[i], p_oot[i])
    g_nuevo.append(gn)
    g_dif.append(gn - gini(y_oot[i], p_ant_oot[i]))

print(f"Gini OOT nuevo:    {gini(y_oot, p_oot):.4f}  IC95% {np.percentile(g_nuevo, [2.5, 97.5]).round(4)}")
print(f"Gini OOT anterior: {gini(y_oot, p_ant_oot):.4f}")
print(f"Mejora IC95%:      {np.percentile(g_dif, [2.5, 97.5]).round(4)}")
print(pd.DataFrame({"nuevo": gini_mensual(p_oot, oot),
                    "anterior": gini_mensual(p_ant_oot, oot)}).round(4))

# 2. Estabilidad del score (PSI DEV vs OOT, por deciles de DEV)
cortes = np.unique(np.quantile(p_dev, np.linspace(0, 1, 11)))
cortes[0], cortes[-1] = -np.inf, np.inf
psi_score = psi(pd.Series(pd.cut(p_dev, cortes, labels=False)),
                pd.Series(pd.cut(p_oot, cortes, labels=False)))
print(f"PSI score: {psi_score:.4f}")

# 3. Calibración
print(f"PD media predicha: {p_oot.mean():.4f} | Tasa observada: {y_oot.mean():.4f}")
calib = pd.DataFrame({"decil": pd.qcut(p_oot, 10, labels=False), "pd": p_oot, "y": y_oot})
print(calib.groupby("decil").agg(pd_media=("pd", "mean"), tasa_obs=("y", "mean"), n=("y", "size")).round(4))
```

```python
SEED = 42
CAT_VARS = [v for v in FEATURES if datos[v].dtype == "object"]
y_dev = dev[TARGET].values

# ---------- 1. Preparación (sin WoE) ----------
def preparar(df, vars_, ref):
    """Variables crudas; categóricas como 'category' con niveles de `ref` (nivel no visto -> NaN)."""
    X = df[vars_].copy()
    for c in vars_:
        if c in CAT_VARS:
            X[c] = pd.Categorical(X[c], categories=sorted(ref[c].dropna().unique()))
    return X

def para_sklearn(X):
    """Árbol/RF de sklearn: categóricas a códigos (-1 = nulo/nuevo), nulos numéricos a centinela."""
    X = X.copy()
    for c in X.columns:
        if isinstance(X[c].dtype, pd.CategoricalDtype):
            X[c] = X[c].cat.codes
    return X.fillna(-999999)

DIR = diag.set_index("variable")["direccion_datos"]
def restricciones(vars_):
    """+1 / -1 según la dirección observada en el binning; 0 para categóricas o planas."""
    return [0 if v in CAT_VARS else {"ascendente": 1, "descendente": -1}.get(DIR[v], 0)
            for v in vars_]

# ---------- 2. Modelos ----------
def ajustar(nombre, hp, Xtr, ytr, Xva=None, yva=None, n_est=None):
    """Con Xva: early stopping en VALID (boosting). Sin Xva: n_est fijo (reentreno en DEV)."""
    if nombre in ("Arbol", "RandomForest"):
        m = (DecisionTreeClassifier(random_state=SEED, **hp) if nombre == "Arbol" else
             RandomForestClassifier(n_estimators=300, max_samples=0.5, n_jobs=-1,
                                    random_state=SEED, **hp))
        return m.fit(para_sklearn(Xtr), ytr), None

    es = Xva is not None
    if nombre == "XGBoost":
        m = xgb.XGBClassifier(n_estimators=3000 if es else n_est, learning_rate=0.03,
                              subsample=0.8, colsample_bytree=0.8, tree_method="hist",
                              enable_categorical=True, eval_metric="auc",
                              early_stopping_rounds=100 if es else None,
                              random_state=SEED, n_jobs=-1, **hp)
        m.fit(Xtr, ytr, eval_set=[(Xva, yva)] if es else None, verbose=False)
        return m, (m.best_iteration + 1 if es else n_est)

    extra = ({"monotone_constraints": restricciones(list(Xtr.columns)),
              "monotone_constraints_method": "advanced"} if nombre == "LightGBM_mono" else {})
    m = lgb.LGBMClassifier(n_estimators=3000 if es else n_est, learning_rate=0.03,
                           subsample=0.8, subsample_freq=1, colsample_bytree=0.8,
                           metric="auc", random_state=SEED, verbose=-1, **extra, **hp)
    if es:
        m.fit(Xtr, ytr, eval_set=[(Xva, yva)],
              callbacks=[lgb.early_stopping(100, verbose=False)])
        return m, m.best_iteration_
    return m.fit(Xtr, ytr), n_est

def predecir(nombre, m, X):
    X = para_sklearn(X) if nombre in ("Arbol", "RandomForest") else X
    return m.predict_proba(X)[:, 1]

def grilla(d):
    return [dict(zip(d, v)) for v in itertools.product(*d.values())]

GRIDS = {
    "Arbol":        grilla({"max_depth": [3, 4, 5, 6, 8], "min_samples_leaf": [0.01, 0.02, 0.05]}),
    "RandomForest": grilla({"max_depth": [6, 8, 12], "min_samples_leaf": [0.002, 0.01],
                            "max_features": ["sqrt", 0.3]}),
    "XGBoost":      grilla({"max_depth": [3, 4, 5], "min_child_weight": [10, 50],
                            "reg_lambda": [1, 10]}),
    "LightGBM":     grilla({"num_leaves": [7, 15, 31], "min_child_samples": [200, 1000],
                            "reg_lambda": [0, 5]}),
}
GRIDS["LightGBM_mono"] = GRIDS["LightGBM"]

SETS = {"17vars": VARS_FINAL, "todas": FEATURES}

# ---------- 3. Tuneo en VALID + reentreno en DEV ----------
BASE = "Logistica WoE | 17vars"
preds_dev, preds_oot = {BASE: p_dev}, {BASE: p_oot}
print(f"{BASE:<28} gini_valid={gini(y_va, p_17):.4f}  (referencia TRAIN->VALID)")

tuneo = []
for set_nom, vars_ in SETS.items():
    Xtr, Xva = preparar(train, vars_, train), preparar(valid, vars_, train)
    Xdev, Xoot = preparar(dev, vars_, dev), preparar(oot, vars_, dev)
    for nombre, grid in GRIDS.items():
        t0, mejor = time.time(), None
        for hp in grid:
            m, n_it = ajustar(nombre, hp, Xtr, y_tr, Xva, y_va)
            g = gini(y_va, predecir(nombre, m, Xva))
            if mejor is None or g > mejor["gini_valid"]:
                mejor = {"hp": hp, "n_iter": n_it, "gini_valid": g}
        # Reentreno en DEV con hiperparámetros (y n° de árboles) fijados en VALID
        m_fin, _ = ajustar(nombre, mejor["hp"], Xdev, y_dev, n_est=mejor["n_iter"])
        clave = f"{nombre} | {set_nom}"
        preds_dev[clave] = predecir(nombre, m_fin, Xdev)
        preds_oot[clave] = predecir(nombre, m_fin, Xoot)
        tuneo.append({"modelo": clave, **mejor, "seg": round(time.time() - t0)})
        print(f"{clave:<28} gini_valid={mejor['gini_valid']:.4f}  "
              f"n_iter={mejor['n_iter']}  hp={mejor['hp']}  ({tuneo[-1]['seg']}s)")

# ---------- 4. Evaluación en OOT (misma batería que la logística) ----------
preds_oot["Modelo anterior"] = p_ant_oot
PCT_APROB = (p_oot <= CORTE).mean()   # % de aptos que genera tu política actual en OOT

filas, gm_oot = [], {}
for clave, po in preds_oot.items():
    rng = np.random.default_rng(42)    # mismos remuestreos para todos -> bootstrap pareado
    dif = []
    for _ in range(1000):
        i = rng.integers(0, len(y_oot), len(y_oot))
        dif.append(gini(y_oot[i], po[i]) - gini(y_oot[i], p_oot[i]))

    pdv = preds_dev.get(clave)
    g_dev = gini(y_dev, pdv) if pdv is not None else np.nan
    if pdv is not None:
        c = np.unique(np.quantile(pdv, np.linspace(0, 1, 11)))
        c[0], c[-1] = -np.inf, np.inf
        psi_s = psi(pd.Series(pd.cut(pdv, c, labels=False)), pd.Series(pd.cut(po, c, labels=False)))
    else:
        psi_s = np.nan

    gm_oot[clave] = gini_mensual(po, oot)
    g_oot = gini(y_oot, po)
    filas.append({
        "modelo": clave,
        "gini_dev": g_dev,
        "gini_oot": g_oot,
        "gap_dev_oot": g_dev - g_oot,
        "gini_min_mes": gm_oot[clave].min(),
        "dif_vs_log_li": np.percentile(dif, 2.5),
        "dif_vs_log_ls": np.percentile(dif, 97.5),
        "psi_score": psi_s,
        "pd_media/tasa": po.mean() / y_oot.mean() if pdv is not None else np.nan,
        f"malos_aptos@{PCT_APROB:.0%}": y_oot[po <= np.quantile(po, PCT_APROB)].mean(),
    })

comp = pd.DataFrame(filas).sort_values("gini_oot", ascending=False)
print(comp.round(4).to_string(index=False))
print("\nGini mensual OOT:")
print(pd.DataFrame(gm_oot).T.round(4))
```
